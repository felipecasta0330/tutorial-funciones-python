## 24. `map()`

Aplica una función a **cada elemento** de una colección y devuelve un iterador con los resultados. Es una forma compacta de transformar datos sin escribir un ciclo.

**Sintaxis:** `map(funcion, iterable)`

```python
celsius = [0, 15, 25, 37]
fahrenheit = list(map(lambda c: c * 9 / 5 + 32, celsius))
print(fahrenheit)

# Convertir una lista de textos a números
textos = ["4", "8", "15", "16"]
numeros = list(map(int, textos))
print(numeros, "-> suma:", sum(numeros))

# Limpiar nombres
nombres = ["  ana ", "LUIS", "marta  "]
print(list(map(lambda n: n.strip().title(), nombres)))
```

<a id="filter"></a>
## 25. `filter()`

Devuelve solo los elementos que **cumplen una condición**. La condición es una función que devuelve `True` o `False`.

**Sintaxis:** `filter(funcion, iterable)`

```python
numeros = [5, 12, 7, 20, 3, 18]
pares = list(filter(lambda n: n % 2 == 0, numeros))
print("Pares:", pares)

# Caso práctico: notas aprobadas (>= 3.0)
notas = [2.5, 3.8, 4.2, 2.9, 3.0]
aprobadas = list(filter(lambda n: n >= 3.0, notas))
print("Aprobadas:", aprobadas)

# Con None como función, elimina los valores vacíos
datos = ["a", "", "b", None, "c"]
print(list(filter(None, datos)))
```

<a id="any"></a>
## 26. `any()`

Devuelve `True` si **al menos un** elemento de la colección es verdadero.

**Sintaxis:** `any(iterable)`

```python
print(any([False, False, True]))
print(any([0, 0, 0]))
print(any([]))           # colección vacía -> False

# Caso práctico: ¿hay algún sensor por encima del límite?
lecturas = [22.1, 23.4, 31.7, 24.0]
limite = 30
if any(t > limite for t in lecturas):
    print("Alerta: al menos una lectura supera", limite)
```

<a id="all"></a>
## 27. `all()`

Devuelve `True` si **todos** los elementos de la colección son verdaderos.

**Sintaxis:** `all(iterable)`

```python
print(all([True, True, True]))
print(all([1, 2, 0]))
print(all([]))           # colección vacía -> True

# Caso práctico: validar que un formulario esté completo
formulario = {"nombre": "Ana", "correo": "ana@mail.com", "telefono": ""}
if all(formulario.values()):
    print("Formulario completo")
else:
    print("Faltan campos por llenar")
```

<a id="pow"></a>
## 28. `pow()`

Calcula una **potencia**. Con un tercer argumento calcula la potencia y luego el residuo de la división (muy usado en criptografía).

**Sintaxis:** `pow(base, exponente)` o `pow(base, exponente, modulo)`

```python
print(pow(2, 10))       # 2 elevado a la 10
print(pow(9, 0.5))      # raíz cuadrada
print(pow(2, -2))       # 1 / 2^2
print(pow(3, 4, 5))     # (3^4) % 5

# Caso práctico: interés compuesto
capital = 1_000_000
tasa = 0.08
anios = 5
final = capital * pow(1 + tasa, anios)
print("Capital después de", anios, "años:", round(final))
```

<a id="divmod"></a>
## 29. `divmod()`

Hace una división entera y devuelve **dos valores a la vez**: el cociente y el residuo.

**Sintaxis:** `divmod(a, b)` → `(a // b, a % b)`

```python
print(divmod(17, 5))

# Caso práctico: convertir segundos a minutos y segundos
segundos_totales = 3725
minutos, segundos = divmod(segundos_totales, 60)
horas, minutos = divmod(minutos, 60)
print(f"{horas} h {minutos} min {segundos} s")
```

<a id="ord"></a>
## 30. `ord()`

Devuelve el **código numérico (Unicode)** de un carácter.

**Sintaxis:** `ord(caracter)`

```python
print(ord("A"))
print(ord("a"))
print(ord("ñ"))
print(ord("0"))

# Caso práctico: posición de una letra en el alfabeto
letra = "D"
print(letra, "es la letra número", ord(letra) - ord("A") + 1)
```

<a id="chr"></a>
## 31. `chr()`

Hace lo contrario de `ord()`: recibe un **código numérico** y devuelve el carácter correspondiente.

**Sintaxis:** `chr(codigo)`

```python
print(chr(65))
print(chr(241))

# Generar el abecedario en mayúsculas
abecedario = [chr(c) for c in range(ord("A"), ord("Z") + 1)]
print("".join(abecedario))

# Caso práctico: cifrado César sencillo (desplazar 3 letras)
mensaje = "HOLA"
cifrado = "".join(chr((ord(l) - 65 + 3) % 26 + 65) for l in mensaje)
print(mensaje, "->", cifrado)
```

<a id="format"></a>
## 32. `format()`

Da **formato a un valor** según una especificación: cantidad de decimales, separador de miles, porcentaje, ancho, alineación, etc.

**Sintaxis:** `format(valor, especificacion)`

```python
print(format(3.14159, ".2f"))        # 2 decimales
print(format(1234567.891, ",.2f"))   # separador de miles
print(format(0.856, ".1%"))          # porcentaje
print(format(42, "08d"))             # rellenar con ceros
print(format("Ana", ">10"))          # alinear a la derecha

# Caso práctico: tabla alineada
productos = [("Café", 3500), ("Queso", 9800), ("Pan", 1200)]
for nombre, precio in productos:
    print(format(nombre, "<8"), format(precio, ">8,"))
```

<a id="dir"></a>
## 33. `dir()`

Lista los **atributos y métodos** disponibles para un objeto. Sirve para descubrir qué se puede hacer con un tipo de dato.

**Sintaxis:** `dir(objeto)`

```python
# Métodos de una lista (se omiten los que empiezan con __)
metodos_lista = [m for m in dir([]) if not m.startswith("_")]
print(metodos_lista)

# Métodos de un texto que empiezan por "is"
metodos_texto = [m for m in dir("") if m.startswith("is")]
print(metodos_texto)
```

## 35. `open()`

Abre un **archivo** para leerlo o escribir en él. Se recomienda usarlo con `with`, que cierra el archivo automáticamente.

**Sintaxis:** `open(ruta, modo, encoding)`

Modos más usados: `"r"` leer, `"w"` escribir (borra el contenido anterior), `"a"` agregar al final.

```python
# Escribir un archivo de texto
with open("ejemplo_open.txt", "w", encoding="utf-8") as archivo:
    archivo.write("fecha,temperatura\n")
    archivo.write("2026-10-30,17.5\n")
    archivo.write("2026-10-31,18.2\n")

# Agregar una línea al final
with open("ejemplo_open.txt", "a", encoding="utf-8") as archivo:
    archivo.write("2026-11-01,16.9\n")

# Leerlo línea por línea
with open("ejemplo_open.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea.strip())
```
