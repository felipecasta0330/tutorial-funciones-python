## 1. `print()`

Muestra uno o varios valores en pantalla. Es la forma más básica de ver resultados mientras se programa.

**Sintaxis:** `print(valor1, valor2, ..., sep=' ', end='\n')`

- `sep`: texto que se pone entre los valores (por defecto, un espacio).
- `end`: texto que se pone al final (por defecto, un salto de línea).

```python
temperatura = 18.5
ciudad = "Bogotá"

print("Hola, mundo")
print("Temperatura en", ciudad, ":", temperatura, "°C")

# sep cambia el separador entre valores
print("2026", "10", "30", sep="-")

# end evita el salto de línea
print("Cargando", end="... ")
print("listo")

# f-strings: forma recomendada de mezclar texto y variables
print(f"En {ciudad} hay {temperatura} °C")
```

<a id="type"></a>
## 2. `type()`

Devuelve el **tipo de dato** de un valor. Es muy útil para revisar cómo quedó guardada una variable antes de operar con ella (por ejemplo, si un número llegó como texto).

**Sintaxis:** `type(objeto)`

```python
print(type(25))
print(type(3.1416))
print(type("25"))
print(type(True))
print(type([1, 2, 3]))
print(type({"nombre": "Ana"}))

# Caso práctico: un dato leído de un archivo suele llegar como texto
edad_leida = "21"
print("¿Es texto?", type(edad_leida) == str)
```

<a id="int"></a>
## 3. `int()`

Convierte un valor a **número entero**. Si recibe un decimal, lo **trunca** (corta la parte decimal, no redondea). Si recibe texto, el texto debe representar un entero válido.

**Sintaxis:** `int(valor)` o `int(texto, base)`

```python
print(int("42") + 8)        # texto -> entero
print(int(7.9))              # trunca: 7, no 8
print(int(-7.9))             # trunca hacia cero: -7
print(int(True))             # True vale 1

# Con base: convertir un número binario a decimal
print(int("1010", 2))

# Caso práctico: sumar cantidades que llegaron como texto
cantidades = ["3", "10", "7"]
total = 0
for c in cantidades:
    total = total + int(c)
print("Total de unidades:", total)
```

<a id="float"></a>
## 4. `float()`

Convierte un valor a **número decimal** (punto flotante). Acepta enteros y textos con formato numérico, incluyendo notación científica.

**Sintaxis:** `float(valor)`

```python
print(float("3.75"))
print(float(10))
print(float("1e-3"))     # notación científica
print(float("  2.5  "))  # ignora espacios al inicio y al final

# Caso práctico: los datos con coma decimal deben arreglarse primero
precio_texto = "12500,50"
precio = float(precio_texto.replace(",", "."))
print("Precio:", precio)
```

<a id="str"></a>
## 5. `str()`

Convierte cualquier valor a **texto** (cadena de caracteres). Se usa para construir mensajes o guardar números como texto.

**Sintaxis:** `str(objeto)`

```python
numero = 2026
texto = str(numero)
print(texto, type(texto))

# Concatenar texto con un número requiere convertirlo
print("Año: " + str(numero))

print(str(3.5))
print(str([1, 2, 3]))
print(str(None))
```

<a id="bool"></a>
## 6. `bool()`

Convierte un valor a **booleano** (`True` o `False`). En Python, los valores "vacíos" o cero se consideran `False`: `0`, `0.0`, `""`, `[]`, `{}` y `None`. Todo lo demás es `True`.

**Sintaxis:** `bool(valor)`

```python
print(bool(1), bool(0))
print(bool("hola"), bool(""))
print(bool([1, 2]), bool([]))
print(bool(None))

# Caso práctico: validar que un campo no esté vacío
nombre = ""
if not bool(nombre):
    print("El campo 'nombre' está vacío")
```

<a id="len"></a>
## 7. `len()`

Devuelve la **cantidad de elementos** de una secuencia o colección: caracteres de un texto, elementos de una lista, claves de un diccionario, etc.

**Sintaxis:** `len(objeto)`

```python
print(len("Uniagraria"))
print(len([10, 20, 30, 40]))
print(len({"a": 1, "b": 2}))
print(len(""))

# Caso práctico: contar registros y calcular un promedio
notas = [4.2, 3.8, 4.5, 3.9, 4.0]
promedio = sum(notas) / len(notas)
print("Número de notas:", len(notas))
print("Promedio:", promedio)
```

<a id="round"></a>
## 8. `round()`

Redondea un número a una cantidad de **cifras decimales**. Si no se indican decimales, devuelve un entero.

**Sintaxis:** `round(numero, ndigitos)`

**Ojo:** Python usa el "redondeo del banquero": cuando el número está exactamente en la mitad, redondea al par más cercano (`round(2.5)` da `2`).

```python
print(round(3.14159, 2))
print(round(3.7))
print(round(1234.5678, -2))   # negativo: redondea a centenas

print(round(2.5), round(3.5))  # redondeo al par

# Caso práctico: mostrar un porcentaje legible
aprobados, total = 17, 23
porcentaje = aprobados / total * 100
print("Porcentaje de aprobación:", round(porcentaje, 1), "%")
```

<a id="abs"></a>
## 9. `abs()`

Devuelve el **valor absoluto** de un número, es decir, su distancia al cero sin importar el signo.

**Sintaxis:** `abs(numero)`

```python
print(abs(-15))
print(abs(15))
print(abs(-3.5))

# Caso práctico: error entre un valor medido y uno esperado
esperado = 25.0
medido = 23.8
error = abs(medido - esperado)
print("Error absoluto:", round(error, 2))
```

<a id="isinstance"></a>
## 10. `isinstance()`

Comprueba si un valor es de un tipo determinado y devuelve `True` o `False`. A diferencia de comparar con `type()`, permite revisar **varios tipos a la vez** pasando una tupla.

**Sintaxis:** `isinstance(objeto, tipo)` o `isinstance(objeto, (tipo1, tipo2))`

```python
print(isinstance(10, int))
print(isinstance(10, float))
print(isinstance(10.5, (int, float)))

# Caso práctico: quedarse solo con los datos numéricos de una lista mezclada
datos = [12, "N/A", 15.5, None, 9, "error", 20]
numericos = []
for d in datos:
    if isinstance(d, (int, float)):
        numericos.append(d)
print("Datos válidos:", numericos)
```

<a id="input"></a>
## 11. `input()`

Pide al usuario que escriba un dato por teclado y lo devuelve **siempre como texto** (`str`). Por eso, si se necesita un número, hay que convertirlo con `int()` o `float()`.

**Sintaxis:** `input(mensaje)`

**Nota:** al ejecutar la celda, Jupyter muestra un cuadro para escribir. En este notebook el ejemplo está en una función para poder probarlo sin bloquear la ejecución del resto.

```python
def pedir_edad():
    edad_texto = input("Escribe tu edad: ")   # siempre devuelve texto
    edad = int(edad_texto)                     # se convierte a número
    print("El próximo año tendrás", edad + 1, "años")

# Para probarla, quita el # de la siguiente línea y ejecuta la celda:
# pedir_edad()

print("input() devuelve:", type("21"), "-> hay que convertirlo con int() o float()")
```
