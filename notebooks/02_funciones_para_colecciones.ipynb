## 12. `list()`

Crea una **lista**: una colección ordenada y modificable. También convierte otras secuencias (texto, tuplas, rangos, conjuntos) en lista.

**Sintaxis:** `list()` o `list(iterable)`

```python
vacia = list()
print(vacia)

print(list("datos"))           # cada carácter se vuelve un elemento
print(list((1, 2, 3)))          # tupla -> lista
print(list(range(5)))           # rango -> lista

# Las listas se pueden modificar
frutas = list(("mango", "pera"))
frutas.append("uva")
print(frutas)
```

<a id="tuple"></a>
## 13. `tuple()`

Crea una **tupla**: una colección ordenada pero **inmutable** (no se puede modificar después de creada). Se usa para datos que no deben cambiar, como coordenadas o registros fijos.

**Sintaxis:** `tuple()` o `tuple(iterable)`

```python
coordenada = tuple([4.60971, -74.08175])
print(coordenada)
print("Latitud:", coordenada[0])

print(tuple("abc"))

# Intentar modificarla produce un error
try:
    coordenada[0] = 0
except TypeError as e:
    print("Error:", e)
```

<a id="set"></a>
## 14. `set()`

Crea un **conjunto**: una colección **sin elementos repetidos** y sin orden. Es la forma más rápida de eliminar duplicados o de comparar grupos.

**Sintaxis:** `set()` o `set(iterable)`

```python
ciudades = ["Bogotá", "Cali", "Bogotá", "Medellín", "Cali"]
unicas = set(ciudades)
print(unicas)
print("Ciudades distintas:", len(unicas))

# Operaciones entre conjuntos
grupo_a = set(["Ana", "Luis", "Marta"])
grupo_b = set(["Luis", "Pedro"])
print("En ambos grupos:", grupo_a & grupo_b)
print("En cualquiera:", grupo_a | grupo_b)
print("Solo en A:", grupo_a - grupo_b)
```

<a id="dict"></a>
## 15. `dict()`

Crea un **diccionario**: una colección de pares **clave: valor**. Permite buscar información por un nombre en lugar de por posición.

**Sintaxis:** `dict()`, `dict(clave=valor, ...)` o `dict(lista_de_pares)`

```python
estudiante = dict(nombre="Ana", semestre=6, promedio=4.1)
print(estudiante)
print("Promedio:", estudiante["promedio"])

# A partir de una lista de pares
precios = dict([("café", 3500), ("pan", 1200)])
print(precios)

# Agregar y consultar con valor por defecto
precios["leche"] = 4800
print(precios.get("queso", "no disponible"))
```

<a id="range"></a>
## 16. `range()`

Genera una **secuencia de números enteros**. Se usa sobre todo para repetir algo un número de veces dentro de un ciclo `for`.

**Sintaxis:** `range(fin)`, `range(inicio, fin)` o `range(inicio, fin, paso)`

El valor `fin` **no se incluye**.

```python
print(list(range(5)))           # 0 a 4
print(list(range(2, 8)))        # 2 a 7
print(list(range(0, 20, 5)))    # de 5 en 5
print(list(range(10, 0, -2)))   # hacia atrás

# Caso práctico: tabla de multiplicar
for i in range(1, 6):
    print(f"7 x {i} = {7 * i}")
```

<a id="sorted"></a>
## 17. `sorted()`

Devuelve una **nueva lista ordenada** a partir de cualquier colección, sin modificar la original.

**Sintaxis:** `sorted(iterable, key=None, reverse=False)`

- `reverse=True`: orden de mayor a menor.
- `key`: función que indica **por qué criterio** ordenar.

```python
numeros = [42, 7, 19, 3, 25]
print(sorted(numeros))
print(sorted(numeros, reverse=True))
print("Original sin cambios:", numeros)

palabras = ["kiwi", "banano", "uva", "mandarina"]
print(sorted(palabras))              # alfabético
print(sorted(palabras, key=len))     # por longitud

# Caso práctico: ordenar estudiantes por nota
notas = [("Ana", 4.1), ("Luis", 3.6), ("Marta", 4.7)]
print(sorted(notas, key=lambda x: x[1], reverse=True))
```

<a id="reversed"></a>
## 18. `reversed()`

Devuelve un **iterador** que recorre una secuencia **al revés**. Para ver el resultado completo se puede convertir a lista.

**Sintaxis:** `reversed(secuencia)`

```python
numeros = [1, 2, 3, 4, 5]
print(list(reversed(numeros)))

# Recorrer en orden inverso
for n in reversed(numeros):
    print(n, end=" ")
print()

# Invertir un texto
palabra = "python"
print("".join(reversed(palabra)))
```

<a id="enumerate"></a>
## 19. `enumerate()`

Recorre una colección entregando **al mismo tiempo el índice y el valor** de cada elemento. Evita tener que llevar un contador manual.

**Sintaxis:** `enumerate(iterable, start=0)`

```python
tareas = ["Leer guía", "Escribir código", "Hacer commit"]

for i, tarea in enumerate(tareas):
    print(i, tarea)

print()
# start=1 para numerar desde 1
for i, tarea in enumerate(tareas, start=1):
    print(f"{i}. {tarea}")
```

<a id="zip"></a>
## 20. `zip()`

Combina **varias colecciones elemento por elemento**, como si se pegaran columnas lado a lado. Se detiene cuando la colección más corta se acaba.

**Sintaxis:** `zip(iterable1, iterable2, ...)`

```python
nombres = ["Ana", "Luis", "Marta"]
edades = [21, 23, 22]
ciudades = ["Bogotá", "Cali", "Tunja"]

for n, e, c in zip(nombres, edades, ciudades):
    print(f"{n} tiene {e} años y vive en {c}")

# Caso práctico: construir un diccionario a partir de dos listas
print(dict(zip(nombres, edades)))
```

<a id="sum"></a>
## 21. `sum()`

Suma todos los elementos numéricos de una colección. Opcionalmente recibe un valor inicial.

**Sintaxis:** `sum(iterable, start=0)`

```python
ventas = [120000, 85000, 230000, 64000]
print("Total vendido:", sum(ventas))
print("Con saldo inicial:", sum(ventas, 50000))

# Caso práctico: promedio
temperaturas = [14.2, 16.8, 19.1, 17.5]
print("Promedio:", round(sum(temperaturas) / len(temperaturas), 2))
```

<a id="max"></a>
## 22. `max()`

Devuelve el **elemento más grande** de una colección, o el mayor entre varios valores. Con `key` se puede indicar el criterio de comparación.

**Sintaxis:** `max(iterable, key=None)` o `max(valor1, valor2, ...)`

```python
print(max([3, 17, 9, 25, 4]))
print(max(8, 2, 14))
print(max(["perro", "gato", "elefante"], key=len))

# Caso práctico: producto más caro de un diccionario
precios = {"café": 3500, "pan": 1200, "queso": 9800}
mas_caro = max(precios, key=precios.get)
print("Más caro:", mas_caro, "->", precios[mas_caro])
```

<a id="min"></a>
## 23. `min()`

Devuelve el **elemento más pequeño** de una colección, o el menor entre varios valores. Funciona igual que `max()`.

**Sintaxis:** `min(iterable, key=None)` o `min(valor1, valor2, ...)`

```python
print(min([3, 17, 9, 25, 4]))
print(min(8, 2, 14))
print(min("python"))     # letra con menor valor alfabético

# Caso práctico: rango de los datos (máximo - mínimo)
datos = [14.2, 16.8, 19.1, 12.7, 17.5]
print("Rango:", round(max(datos) - min(datos), 2))
```
